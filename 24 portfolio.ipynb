```python
import pandas as pd
```

```python
df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['young adult','adult','middle aged','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases']
```

```python

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']]
```

```python
from sqlalchemy import create_engine, text
import pandas as pd

# Step 1: Connect to MySQL server without a database to ensure it exists
temp_engine = create_engine("mysql+mysqlconnector://root:root@127.0.0.1:3306/")
with temp_engine.connect() as conn:
    # Set isolation level to auto-commit so the database gets created immediately
    conn.execution_options(isolation_level="AUTOCOMMIT").execute(text("CREATE DATABASE IF NOT EXISTS customer_behavior;"))

# Step 2: Now connect to the newly created database
username = "root"
password = "root"
host = "127.0.0.1"
port = "3306"
database = "customer_behavior"

engine = create_engine(f"mysql+mysqlconnector://{username}:{password}@{host}:{port}/{database}")

# Step 3: Load your DataFrame into MySQL
table_name = "customer"
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
